```python
%sql
SELECT COUNT(*) AS total_rows,
       COUNT(DISTINCT SalesOrderNumber, ProductKey) AS distinct_keys
FROM fact_sales
```

```python
%sql
SELECT 'fact_sales' AS tbl, COUNT(*) AS n FROM fact_sales
UNION ALL SELECT 'dim_product', COUNT(*) FROM dim_product
UNION ALL SELECT 'dim_sales_territory', COUNT(*) FROM dim_sales_territory
```

```python
%sql
SELECT COUNT(*) FROM fact_sales f LEFT JOIN dim_product p ON f.ProductKey = p.ProductKey
```

```python
%sql
SELECT COUNT(*) FROM fact_sales f LEFT JOIN dim_sales_territory t ON f.SalesTerritoryKey = t.SalesTerritoryKey
```

```python
%sql
SELECT COUNT(*) FROM fact_sales f RIGHT JOIN dim_product p ON f.ProductKey = p.ProductKey
```

### Findings
- The key (SalesOrderNumber + ProductKey) is unique: 57,851 rows = 57,851 distinct keys.
- fact_sales has 57,851 rows. dim_product has 397 and dim_sales_territory has 10, which is plausible for a fact table and two small dimensions.
- LEFT JOINs to both dimensions keep 57,851 rows, so every key matches and there is no fan-out.
- The RIGHT JOIN to dim_product gives 57,914, because it keeps the 63 products that never sold (NULL sales columns).
- No duplicate ProductKey exists in dim_product.
- A duplicate key in a dimension makes each fact row match two dimension rows, so its revenue is counted twice and SUMs inflate with no error.
